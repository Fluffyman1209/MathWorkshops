```python
import numpy as np
def magnitude(v):
    return np.linalg.norm(v)
def direction_cosines(v):
    return v / np.linalg.norm(v)
def scalar_product(a, b):
    return np.dot(a, b)
def vector_product(a, b):
    return np.cross(a, b)
U = np.array([1.5, 1.5, 2.0])
O = np.array([1.5, 1.5, 2.0])
a = np.array([1.5, 1.5, 2.0])
b = np.array([-1.5, 0.5, 2.0])
print(magnitude(U))
print(direction_cosines(O))
print(scalar_product(a, b))
print(vector_product(a, b))
```

```python
import numpy as np
import matplotlib.pyplot as plt
U = np.array([1, 0.25, 0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.quiver(0,0,0, U[0],U[1],U[2], color='black', label='U')
ax.quiver(0,0,0, V[0],V[1],V[2], color='black', label='V')
ax.quiver(0,0,0, W[0],W[1],W[2], color='black', label='W')
ax.set_xlim([0,1.5])
ax.set_ylim([0,1.5])
ax.set_zlim([0,1.5])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
U = np.array([-0.5, 0.5])
V = np.array([1, 1])
W = U + V
plt.quiver(0,0,U[0],U[1],angles='xy',scale_units='xy',scale=1,color='black',label='U')
plt.quiver(0,0,V[0],V[1],angles='xy',scale_units='xy',scale=1,color='black',label='V')
plt.quiver(0,0,W[0],W[1],angles='xy',scale_units='xy',scale=1,color='black',label='W')
plt.xlim(-1,2)
plt.ylim(-1,2)
plt.axhline(0,color='black')
plt.axvline(0,color='black')
plt.legend()
plt.show()
```
